## Number operations
Python supports all basic operations on numbers.
Because certain characters cannot be typed in Python code (like $\div$), the notation is slightly different.

For *multiplication*, instead of:

$\qquad ab\ =\ a\ \times\ b\\\qquad ab\ =\ a\ \cdot\ b$

we say:

$\qquad ab\ =\ a\ \texttt{*}\ b$

For *division*, instead of:

$\qquad\frac{a}{b}\ =\ a\ \div b$

we say:

$\qquad \frac{a}{b}\ =\ a\ /\ b$

For *exponentiation*, the number raised to a power, we say:

$\qquad x^y\ =\ x\ \texttt{**}\ y$

The *modulo* operation gives the remainder of a division, so since:

$\qquad 7 \div 3 = 2\ R\ 1$

we say:

$\qquad 7 \ \%\  3 = 1$



## Operation priority
These operations are applied in a specific order.

|Operation |Symbol | Order |
| - | - | - |
|Parentheses| `()` | 1 |
|Exponentiation| `**` | 2 |
|Negation | `-` | 3 |
|Multiply | `*` | 4 |
|Division | `/` | 4 |
|Modulo |`%` | 4 |
|Addition |`+` | 5 |
|Subtraction |`-` | 5 |

For example, this:

$\qquad 5\ +\ 10\ /\ 5\ -\ 1\ =\ 6$

is the same as:

$\qquad 5\ +\ (10\ /\ 5)\ -\ 1\ =\ 6$

Parentheses $($ and $)$ change the application order, such as:

$\qquad (5\ +\ 10)\ /\ 5\ -\ 1\ =\ 2$

Operations of the same priority are applied left to right, so these are the same:

$\qquad 5\ +\ 2\ -\ 1\ =\ (5\ +\ 2)\ -\ 1\ =\ 6$

It occasionally may be clearer and less error prone to use parentheses even if they not necessary.
