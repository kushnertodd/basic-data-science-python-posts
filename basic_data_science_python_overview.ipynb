## Data Science

[Data science](https://www.ibm.com/think/topics/data-science) uses mathematics and statistics to enable organizations to extract insights and guide operational decisions from raw data.
It combines mathematics, computing and domain knowledge to solve real-world problems and uncover hidden patterns.
Companies are generating vast amounts of data that data scientists can clean, statistically analyze, transform, and visualize to discover patterns and predict events in the future.
These can be presented as actionable insights to business decision makers.

## Why Python?

Python is currently the most popular  programming language in the world, and is the standard for data science because its simplicity makes it easy to learn and use.
It was created by [Guido van Rossum](https://gvanrossum.github.io/) in 1991, named for the BBC comedy series [Monty Python’s Flying Circus](https://docs.python.org/3/faq/general.html#why-is-it-called-python).
It's usefulness for data science is because of the large ecosystem of open-source third-party packages
and the large community of developers who create libraries and tools to make Python easier to use for Data Science.

# Google Colab notebooks
This tutorial uses Google Colab notebooks, which are web pages combining both explanations and code boxes, or *sandboxes*, where you can run and experiment with Python code.
This describes how to use [Colab notebooks](https://basicdatascience.substack.com/p/basic-data-science-google-colab).

The notebooks include explanations and boxes containing Python code.
In the code boxes,
text after `#` are comments for descriptive purposes only and are ignored.
The code in the box can be run three ways by clicking in the code box and choosing one of these actions.

| action | after running the code |
| - | - |
| select the Play button (a triangle inside a circle) on the far-left side of the box | keeps the cursor in the same box
| press Ctrl + Enter (Windows) or Cmd + Enter (Mac) | keeps the cursor in the same box
| press Shift + Enter | moves your cursor down to the next box
| press Alt + Enter (Windows) or Option + Enter (Mac) | inserts a new, empty code box directly below it in which you can add new code

The Python `print()` function is used to print the results of Python code, which is displayed below the box.
If you see a popup saying "Warning: This notebook was not authored by Google," click Run anyway.

# Basic Python Data Types
Python has four basic built-in types of data:
- [Strings](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_strings.ipynb): sequences of text characters
- [Numbers](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_numbers.ipynb): positive and negative numbers, with or without fractions
- [Booleans](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_booleans.ipynb): whether a condition or state is true or not

# Python Data Structures
Python data structures collect data values into a group. These include:
- [Lists](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_lists.ipynb): an ordered sequence of values
- [Sets](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_sets.ipynb): an unordered group of values with no duplicates
- [Dictionaries](https://colab.research.google.com/github/kushnertodd/basic-data-science-python-posts/blob/main/basic_data_science_python_dictionaries.ipynb): an unordered group where values are associated with a key
